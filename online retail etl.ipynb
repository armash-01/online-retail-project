STEP 6 — Create SalesAmount

```python
df["SalesAmount"] = df["Quantity"] * df["UnitPrice"]
```

STEP 7 — Identify Cancelled Invoices


```python
df["InvoiceNo"].astype(str).str.startswith("C").sum()
```

```python
cancelled = df[
    df["InvoiceNo"].astype(str).str.startswith("C")
]
```

```python
print(cancelled.head())
```

```python
df = df[
    ~df["InvoiceNo"].astype(str).str.startswith("C")
]
```

```python
print(
    df["InvoiceNo"]
    .astype(str)
    .str.startswith("C")
    .sum()
)
```

STEP 8 — Handle Negative Quantity

```python
negative_qty = df[df["Quantity"] < 0]
```

```python
print(negative_qty)
```

```python
print((df["Quantity"] < 0).sum())
```

```python
df = df[df["Quantity"] > 0]
```

STEP 9 — Standardize Product Description

```python
print(df["Description"].head(20))
```

```python
df["Description"] = df["Description"].str.strip()
```

```python
df["Description"] = df["Description"].str.upper()
```

```python
print(df["Description"].isnull().sum())
```

```python
df = df.dropna(subset=["Description"])
```

STEP 10 — Create Year, Month and Quarter

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
```

```python
df["Year"] = df["InvoiceDate"].dt.year
```

```python
df["Month"] = df["InvoiceDate"].dt.month
```

```python
df["MonthName"] = df["InvoiceDate"].dt.month_name()
```

```python
df["Quarter"] = df["InvoiceDate"].dt.quarter
```
